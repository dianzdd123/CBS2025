```python
import matplotlib.pyplot as plt

x_labels = ['1×1', '2×2', '4×4', '6×6', '8×8']
x = [1, 2, 3, 4, 5] 
mse = [0.037, 0.035, 0.034, 0.039, 0.052]
r2 =  [0.997, 0.997, 0.998, 0.997, 0.996]

adaptive_x = 3.5   # 位于 4×4 和 6×6 中间
adaptive_mse = 0.027
adaptive_r2 = 0.999

fig, ax1 = plt.subplots(figsize=(8, 4))

color1 = 'tab:blue'
ax1.set_xlabel('Grid Size')
ax1.set_ylabel('MSE', color=color1)
ax1.plot(x, mse, marker='o', color=color1, label='MSE')
ax1.tick_params(axis='y', labelcolor=color1)
ax1.set_xticks(x)
ax1.set_xticklabels(x_labels)

ax2 = ax1.twinx()
color2 = 'tab:red'
ax2.set_ylabel('$R^2$', color=color2)
ax2.plot(x, r2, marker='s', color=color2, label='$R^2$')
ax2.tick_params(axis='y', labelcolor=color2)

ax1.scatter(adaptive_x, adaptive_mse, color=color1, s=120, marker='*', label='MSE: Adaptive Grids')
# ax1.plot(adaptive_x, adaptive_mse, marker='*', markersize=12, color=color1, label='Adaptive Grids (MSE)')
# ax2.plot(adaptive_x, adaptive_r2, marker='*', markersize=12, color=color2, label='Adaptive Grids ($R^2$)')
ax2.scatter(adaptive_x, adaptive_r2, color=color2, s=120, marker='*', label='$R^2$: Adaptive Grids')

fig.legend(loc='upper left',  bbox_to_anchor=(0.12, 0.95))

# plt.title("Comparison of Grid Strategies for Depth Estimation")
plt.grid(True)
fig.tight_layout()

plt.savefig("ablation.jpg", dpi=300, bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 数据
data_sizes = ['100%', '80%', '60%', '40%', '20%']
x = np.arange(len(data_sizes))
width = 0.35

# MSE 数据
mse_adaptive = [0.027, 0.028, 0.035, 0.049, 0.065]
mse_resnet50 = [0.075, 0.084, 0.082, 0.144, 0.174]

# R^2 数据
r2_adaptive = [0.999, 0.997, 0.997, 0.996, 0.995]
r2_resnet50 = [0.994, 0.994, 0.994, 0.990, 0.988]

# 创建图
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8))

# ----------- MSE 图 -----------
ax1.bar(x - width/2, mse_adaptive, width, label='Adaptive Grids', color='tab:blue')
ax1.bar(x + width/2, mse_resnet50, width, label='ResNet50', color='gray')
ax1.set_ylabel('MSE')
ax1.set_xlabel('Data Size')
ax1.set_xticks(x)
ax1.set_xticklabels(data_sizes)
ax1.set_title('(a) MSE vs. Data Size')
ax1.legend(loc='upper left')
ax1.grid(True, axis='y', linestyle='--', alpha=0.5)

# ----------- R² 图 -----------
ax2.bar(x - width/2, r2_adaptive, width, label='Adaptive Grids', color='tab:blue')
ax2.bar(x + width/2, r2_resnet50, width, label='ResNet50', color='gray')
ax2.set_ylabel('$R^2$')
ax2.set_xlabel('Data Size')
ax2.set_xticks(x)
ax2.set_xticklabels(data_sizes)
ax2.set_title('(b) $R^2$ vs. Data Size')
ax2.set_ylim(0.975, 1.0)  # 设置 y 轴范围
ax2.legend(loc='lower left')
ax2.grid(True, axis='y', linestyle='--', alpha=0.5)

# 布局和保存
plt.tight_layout()
plt.savefig("data_size_ablation.jpg", dpi=300, bbox_inches='tight')
plt.show()
```
